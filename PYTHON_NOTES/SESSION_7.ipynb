# Object-Oriented Programming Concepts 

## OOP Overview

- Procedural vs. object‑oriented: procedural mixes data & code; OOP groups them in classes.
- OOP centers on **classes** (blueprints) and **objects** (instances of those blueprints).

## Classes and Objects

- **Class** = template that defines attributes (e.g., name, age) and behavior.
- **Object** = concrete instance created from a class, holding its own values.
- Example: “Car” class defines wheels, engine; “BMW” and “Maruti” are objects.


```python
class Car:
    def __init__(self, brand):
        self.brand = brand

c1 = Car("BMW")
c2 = Car("Maruti")
print(c1.brand, c2.brand)   
```

## Constructors (\__init_\_)

- _\_init_\_ runs automatically when you create an object.
- self is the current object (the instance being created/used).

```python
class Person:
    def __init__(self, firstname,lastname, year_of_birth):
        self.firstname = firstname
        self.lastname = lastname
        self.year_of_birth = year_of_birth
```

## Methods

- Functions defined inside a class are called **methods**.
- Example methods:
  - age() – calculates current age from birth year.
  - sum() – returns a simple sum.
  - \__str_\_() – custom string representation for print(obj).

```python
from datetime import date

class Person:
    def __init__(self, firstname, lastname, year_of_birth):
        self.firstname = firstname
        self.lastname = lastname
        self.year_of_birth = year_of_birth

    def age(self):
        return date.today().year - self.year_of_birth

    def full_name(self):
        return f"{self.firstname} {self.lastname}"
   
    def born_info(self):
        return f"{self.full_name()} born in {self.year_of_birth}"


# example
p = Person("Akhila", "Locan", 1881)
print(p.full_name())
print(p.age())
print(p.born_info()) 
```

```python
from datetime import date

class Person:
    def __init__(self, firstname, lastname, year_of_birth):
        self.firstname = firstname
        self.lastname = lastname
        self.year_of_birth = year_of_birth

    def age(self):
        return date.today().year - self.year_of_birth

    def full_name(self):
        return f"{self.firstname} {self.lastname}"

# initialize (positional)
p1 = Person("Alec", "Baldwin", 1958)

# initialize (keyword)
p2 = Person(firstname="Emma", lastname="Watson", year_of_birth=1990)

print(p1.full_name(), p1.age())
print(p2.full_name(), p2.age())
```

## \__str_\_ vs. \__repr_\_

- \__str_\_ → user‑friendly output (e.g., “Alec Baldwin was born in 1958”).
- \__repr_\_ → developer‑friendly, shows internal structure (e.g., Person(name='Alec', ...)).


```python
class Person:
    def __init__(self, firstname, lastname, year_of_birth):
        self.firstname = firstname
        self.lastname = lastname
        self.year_of_birth = year_of_birth

    def full_name(self):
        return f"{self.firstname} {self.lastname}"

    # user-friendly (print(obj))
    def __str__(self):
        return f"{self.full_name()} was born in {self.year_of_birth}"

    # developer-friendly (debugging, list/console)
    def __repr__(self):
        return (f"Person(firstname={self.firstname!r}, "
                f"lastname={self.lastname!r}, "
                f"year_of_birth={self.year_of_birth})")


p = Person("Alec", "Baldwin", 1958)

print(p)      # uses __str__
print([p])    # uses __repr__ inside lists
```

## Abstraction & Name Mangling

- Hide internal details to protect data.
- Single underscore (\_var) signals “private” by convention.
- Double underscore (\__var) triggers name mangling: Python rewrites it to \_ClassName__var, making accidental access harder.

```python
class Person:
    def __init__(self, firstname, lastname, year_of_birth):
        self.firstname = firstname
        self.lastname = lastname
        self.year_of_birth = year_of_birth

        self._nickname = "Aki"       # single underscore: "private" by convention
        self.__ssn = "123-45-6789"   # double underscore: name mangling happens

p = Person("Akhila", "Locan", 1881)

print(p._nickname)       # works (but convention says: don't access from outside)

#print(p.__ssn)         
print(p._Person__ssn)    
```

## Inheritance

- Child class inherits attributes/methods from a parent class, avoiding duplication.
- Use super().\__init_\_(…) to call the parent’s constructor.
- Example: Student inherits from Person and adds student_id.

```python
class Person:
    def __init__(self, firstname, lastname, year_of_birth):
        self.firstname = firstname
        self.lastname = lastname
        self.year_of_birth = year_of_birth

    def full_name(self):
        return f"{self.firstname} {self.lastname}"


class Student(Person):  # Student inherits from Person
    def __init__(self, firstname, lastname, year_of_birth, student_id):
        super().__init__(firstname, lastname, year_of_birth)  # call parent constructor
        self.student_id = student_id  # extra field only for Student

    def student_info(self):
        return f"{self.full_name()} | ID: {self.student_id}"


s = Student("Akhila", "Locan", 1881, "S101")
print(s.full_name())       # inherited method
print(s.student_info())    # child method
```

## Method Overriding

- Child class can redefine a parent method to extend or change behavior.
- Keeps original logic via super() and adds extra steps.

```python
class Person:
    def __init__(self, firstname, lastname, year_of_birth):
        self.firstname = firstname
        self.lastname = lastname
        self.year_of_birth = year_of_birth

    def full_name(self):
        return f"{self.firstname} {self.lastname}"

    def info(self):  # parent method
        return f"{self.full_name()} was born in {self.year_of_birth}"


class Student(Person):
    def __init__(self, firstname, lastname, year_of_birth, student_id):
        super().__init__(firstname, lastname, year_of_birth)
        self.student_id = student_id

    def info(self):  # overriding parent method
        base = super().info()  # keep original logic
        return base + f" and has student ID {self.student_id}"  # add extra


s = Student("Akhila", "Locan", 1881, "101")
print(s.info())
```

## Positional vs. Keyword Arguments

- **Positional**: values passed in order (func(a, b, c)).
- **Keyword**: values passed by name (func(name='A', age=30)) – order doesn’t matter.
- \*args collects extra positional arguments; \*\*kwargs collects extra keyword arguments.

```python
class Person:
    def __init__(self, *args):
        # args must be: (firstname, lastname, year_of_birth)
        self.firstname = args[0]
        self.lastname = args[1]
        self.year_of_birth = args[2]

    def info(self):
        return f"{self.firstname} {self.lastname} born in {self.year_of_birth}"


p = Person("Akhila", "Locan", 1881)
print(p.info())
```

```python
class Person:
    def __init__(self, **kwargs):
        self.firstname = kwargs["firstname"]
        self.lastname = kwargs["lastname"]
        self.year_of_birth = kwargs["year_of_birth"]

    def info(self):
        return f"{self.firstname} {self.lastname} born in {self.year_of_birth}"


p = Person(firstname="Akhila", lastname="Locan", year_of_birth=1881)
print(p.info())
```

```python
class Person:
    def __init__(self, *args, **kwargs):
        if args:
            self.firstname, self.lastname, self.year_of_birth = args
        else:
            self.firstname = kwargs["firstname"]
            self.lastname = kwargs["lastname"]
            self.year_of_birth = kwargs["year_of_birth"]

    def info(self):
        return f"{self.firstname} {self.lastname} born in {self.year_of_birth}"


p1 = Person("Akhila", "Locan", 1881)  # args
p2 = Person(firstname="Akhila", lastname="Locan", year_of_birth=1881)  # kwargs

print(p1.info())
print(p2.info())
```
