## Looping through a list

- for goes one element at a time.
- But the list still already exists in memory.

```python
x = [1,2,3,4,5,6]
for i in x:
    print(i)
```

# Iterator (iter + next)

```python
it = iter(x)      # creates iterator
next(it)          # gives first element
next(it)          # gives next element
...
```

# Generator (yield) — memory efficient
- A generator produces values one at a time without storing the full list.


```python
def genfibo(n):
    a, b = 1, 1
    for _ in range(n):
        yield a
        a, b = b, a + b
```

```python
for num in genfibo(5):
    print(num)
```

	Doesn’t store all numbers in memory
	Generates the next value only when needed

## Iterator vs Generator
	•	Iterator: you can create it using iter() on a collection
	•	Generator: you write it using yield (Python creates iterator behavior for you)

# map — apply function to each element

Instead of writing a loop, map() applies a function to each item.

```python
 lst = [1,2,3,4,5]
result = list(map(lambda x: x+1, lst))
```

```python
list(map(lambda x: x*x, lst))
```

# Reduce — many values → single value

reduce() combines all elements into one final result.

```python
from functools import reduce
lst = [1,2,3,4,5,6]
total = reduce(lambda a,b: a+b, lst)
```

# Filter — keep only matching values

filter() keeps values for which a function returns True.

```python
lst = [1,2,3,4,5,6]
evens = list(filter(lambda x: x % 2 == 0, lst))
```

```python
print(evens)
```
