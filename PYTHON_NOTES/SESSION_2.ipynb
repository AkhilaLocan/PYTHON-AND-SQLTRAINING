# Python Loops: For vs While 

## Loop Types Overview

- Entry‑control loops (e.g., while) evaluate the condition before each iteration.
- Exit‑control loops (e.g., for) know the start and end points up front.
- Use while when the number of iterations isn’t known beforehand.
- Use for when you have a definite sequence or range to traverse.

## For Loop Details

- Iterates over any iterable: list, tuple, string, etc.
- Syntax: for item in iterable: – no need for explicit index unless required.
- Ideal for summing a known list of numbers or concatenating strings.
- Can also loop over indices using range(len(sequence)) for index‑based access.
- The loop stops automatically after the last element; no manual break needed.

## While Loop Details

- Repeats as long as a Boolean condition remains True.
- Suitable for scenarios like “run until a user stops the process.”
- Must update a counter or change the condition inside the loop to avoid infinite loops.
- Example pattern: initialize i = 1, loop while i &lt;= n: …; i += 1.

## Range and Length Functions

- range(start, stop, step) generates a sequence of numbers; stop is **exclusive**.
- Default start is 0; default step is 1.
- Use len(sequence) to get the number of elements; combine with range for dynamic indexing: for i in range(len(seq)):.
- Add +1 to the upper bound when you need to include the last index/value.
- Negative step values produce reverse sequences (e.g., range(8, -1, -2) → 8, 6, 4, 2, 0).

## Common Loop Patterns

- **Summation**:

  ```python
  total = 0
  for num in numbers:
      total += num
  ```

- **Reverse a string**:

  ```python
  rev = ""
  for i in range(len(s)):
      rev = s[i] + rev
  ```

- **Countdown**:

  ```python
  i = 5
  while i >= 0:
      print(i)
      i -= 1
  ```

- **Dynamic iteration**: use range(len(seq)) so the loop adapts to any list length.

## Break and Continue Statements

- break exits the current loop immediately when a condition is met.
- continue skips the rest of the current iteration and proceeds to the next one.
- Typical use: stop iterating once a target value is found, or skip unwanted items.
- Can be combined with while True input loops to exit after a specific user entry.
