# Python Modules, File I/O, and SQLite 

## Modules and Libraries

- Built‑in Python libraries (e.g., math, datetime, os, statistics) provide ready‑made functions.
- Importing a library gives direct access to its functions without writing the underlying logic.
- Use dir(module) or help(module) to explore available functions.
- Custom libraries are just .py files placed in the project or a shared repository.



```python
import math
import datetime
import statistics
import os

print(math.sqrt(49))                 
print(datetime.date.today())         
print(statistics.mean([10, 20, 30])) 
print(os.getcwd())                   

# Explore a module
print(dir(math))         # list of names in math
help(statistics.mean)    # docs for a function
```

## Custom Modules

- Create a [mymath.py](http://mymath.py) file with functions like add, subtract, multiply, divide.
- Import the module (import mymath) and call its functions via mymath.add(...).
- Modules are recognized by the .py extension and can include exception handling.

```python
# mymath.py
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        raise ZeroDivisionError("Cannot divide by zero")
    return a / b
```

## Error and Exception Handling

- Wrap risky code in try: blocks; handle specific errors with except &lt;ErrorType&gt;:.
- Example: catching IOError when opening a file that may not exist.
- Use else: for code that runs only if no exception occurs.
- finally: runs cleanup code (e.g., closing a file) regardless of success or failure.

```python
file = None

try:
    file = open("notes.txt", "r")
    content = file.read()
except FileNotFoundError:
    print("File not found!")
else:
    print("File read successfully:")
    print(content)
finally:
    if file:
        file.close()
        print("Cleanup done (file closed).")
```

```python
try:
    n = int(input("Enter a number: "))
    print(100 / n)
except ValueError:
    print("You must enter a valid integer.")
except ZeroDivisionError:
    print("You entered 0, division not possible.")
```

## File Operations

- Open files with modes: w (write, overwrites), a (append), r (read), r+ (read/write).
- Write multiple lines using \\n for newlines.
- Read whole content with read(), line‑by‑line with a for loop, or all lines as a list with readlines().
- seek(position) moves the file cursor to a specific byte offset.


```python
# Write (overwrites)
with open("sample.txt", "w") as f:
    f.write("Line 1\n")
    f.write("Line 2\n")

# Append
with open("sample.txt", "a") as f:
    f.write("Line 3 (appended)\n")

# Read whole file
with open("sample.txt", "r") as f:
    print(f.read())

# Read line by line
with open("sample.txt", "r") as f:
    for line in f:
        print("->", line.strip())

# readlines() gives list of lines
with open("sample.txt", "r") as f:
    lines = f.readlines()
    print(lines)

# seek() example
with open("sample.txt", "r") as f:
    f.seek(0)          # go to start
    print(f.read(6))   # read first 6 characters
```

## Working with CSV/Excel via pandas

- Load JSON‑like data into a DataFrame: df = pd.DataFrame(data).
- Write DataFrame to Excel: df.to_excel('file.xlsx', index=False) – index=False omits the row index column.
- Write to CSV with df.to_csv('file.csv', index=False).
- Append data by adding rows to the DataFrame then re‑exporting.

```python
import pandas as pd

data = [
    {"name": "Akhila", "score": 92},
    {"name": "Sai", "score": 88},
    {"name": "John", "score": 75},
]

df = pd.DataFrame(data)
print(df)

# Export
df.to_csv("students.csv", index=False)
df.to_excel("students.xlsx", index=False)
```

## Database Connectivity with SQLite

- Import with import sqlite3 and connect: conn = sqlite3.connect('mydb.db').
- Create tables and execute SQL using conn.execute(sql_statement).
- Commit changes with conn.commit(); query with SELECT ... statements.
- The SQLite file is stored locally (e.g., under the C: drive) and initialized quickly.

```python
import sqlite3

# 1) Connect (creates the file if it doesn't exist)
conn = sqlite3.connect("students.db")
cur = conn.cursor()

# 2) Create a table
cur.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    marks INTEGER NOT NULL
)
""")

# 3) Insert data
cur.execute("INSERT INTO students (name, marks) VALUES (?, ?)", ("Akhila", 90))
cur.execute("INSERT INTO students (name, marks) VALUES (?, ?)", ("locan", 85))
conn.commit()

# 4) Read data
cur.execute("SELECT id, name, marks FROM students")
rows = cur.fetchall()

for r in rows:
    print(r)   # (1, 'Akhila', 90)

# 5) Close
conn.close()
```
